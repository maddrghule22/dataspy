## 1. Load Data
Let's import our data loader and read the datasets.

```python
import sys
sys.path.append('../')
from src.data_loader import load_data
train_df, test_df = load_data(data_dir="../dataset")
```

## 2. Outlier Removal
In our EDA, we observed houses with large above-grade living areas (`GrLivArea` > 4000 sq ft) that had surprisingly low sale prices. These are outliers recommended for removal by the dataset creator. We import `remove_outliers` from our preprocessor module.

```python
from src.preprocessing import remove_outliers

print(f"Initial train size: {len(train_df)}")
train_cleaned = remove_outliers(train_df)
print(f"Cleaned train size: {len(train_cleaned)}")
```

## 3. Modular Preprocessor Execution
We'll test our modular `AmesHousingPreprocessor` class, which handles:
- Neighborhood-specific median imputation for `LotFrontage`
- Replacing `NA` with `'None'` or `0` depending on the variable category
- Mode/median imputation for genuine missing values
- Mapping ordinal variables to integers
- Performing One-Hot encoding on nominal variables aligning columns accurately between train and test datasets.

```python
from src.preprocessing import AmesHousingPreprocessor

X = train_cleaned.drop(columns=['SalePrice'])
y = train_cleaned['SalePrice']

# Instantiate preprocessor
preprocessor = AmesHousingPreprocessor()

# Fit preprocessor
preprocessor.fit(X)

# Transform features
X_preprocessed = preprocessor.transform(X)
print(f"Preprocessed train features shape: {X_preprocessed.shape}")
```

```python
# Verify that there are no remaining missing values in the processed set
remaining_nulls = X_preprocessed.isnull().sum().sum()
print(f"Remaining missing values after preprocessing: {remaining_nulls}")
```
